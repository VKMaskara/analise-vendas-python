# Estudos Sobre Pandas

## O que é Pandas?

É uma biblioteca de software de código aberto criada para a linguagem Python, usada principalmente para manipulação, limpeza e análise de dados

Objetivo é facilitar a manipulação de dados e análises, ela foi criada com base na dor de cientistas de dados, analistas e programadores em lidar com dados tabulares de maneira intuitiva e eficiente -> Baseada em Numpy


## Series:
Uma etrutura unidimencional parecida com uma lista, mas com rótulo em casa elemento.

## DataFrame:
Uma estrutura bidimensional, semelhante a uma planilha do Excel ou a uma tabela de banco de dados onde os dados são organixados em linhas e colunas

```python
# Garantir que a data esteja no formato correto
df_vendas['data_venda'] = pd.to_datetime(df_vendas['data_venda'])
```

```python
# ===== AGREGAÇÕES =====

# Total de vendas por categoria
vendas_por_categoria = df_vendas.groupby('categoria')['valor_venda'].sum().reset_index()
print("Total de vendas por categoria:")
print(vendas_por_categoria)
```

```python
# Quantidade de produtos vendidos por categoria
qtd_por_categoria = df_vendas.groupby('categoria')['produto'].count().reset_index()
qtd_por_categoria.columns = ['categoria', 'quantidade']
print("\nQuantidade de vendas por categoria:")
print(qtd_por_categoria)
```

```python
# Média de valor por categoria
media_por_categoria = df_vendas.groupby('categoria')['valor_venda'].mean().reset_index()
print("\nMédia de valor por categoria:")
print(media_por_categoria)
```

```python
# Total de vendas por mês
df_vendas['mes'] = df_vendas['data_venda'].dt.to_period('M')
vendas_por_mes = df_vendas.groupby('mes')['valor_venda'].sum().reset_index()
print("\nTotal de vendas por mês:")
print(vendas_por_mes)
```

```python
# Vou começar a implementar a parte grafica

import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# --- Gráfico 1: Total de vendas por categoria (barra)
plt.figure()
sns.barplot(data=vendas_por_categoria, x='categoria', y='valor_venda', palette='viridis')
plt.title('Total de Vendas por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Valor Total (R$)')
plt.tight_layout()
plt.show()
```

```python
# --- Gráfico 2: Quantidade de vendas por categoria

plt.figure()
sns.barplot(data=qtd_por_categoria, x='categoria', y='quantidade', palette='magma')
plt.title('Quantidade de Vendas por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Quantidade')
plt.tight_layout()
plt.show()
```

```python
# --- Gráfico 3: Evolução das vendas ao longo do ano (linha)
plt.figure()
sns.lineplot(data=vendas_por_mes, x=vendas_por_mes['mes'].astype(str), y='valor_venda', marker='o')
plt.title('Evolução das Vendas ao Longo de 2023')
plt.xlabel('Mês')
plt.ylabel('Valor Total (R$)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# --- Gráfico 4: Distribuição dos valores de venda (boxplot)
plt.figure()
sns.boxplot(data=df_vendas, x='categoria', y='valor_venda', palette='Set2')
plt.title('Distribuição dos Valores de Venda por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Valor da Venda (R$)')
plt.tight_layout()
plt.show()
```

```python
print("\n===== INSIGHTS =====")
print(f"Categoria com maior faturamento: {vendas_por_categoria.loc[vendas_por_categoria['valor_venda'].idxmax(), 'categoria']}")
print(f"Categoria com mais vendas (quantidade): {qtd_por_categoria.loc[qtd_por_categoria['quantidade'].idxmax(), 'categoria']}")
print(f"Mês com maior faturamento: {vendas_por_mes.loc[vendas_por_mes['valor_venda'].idxmax(), 'mes']}")
```
